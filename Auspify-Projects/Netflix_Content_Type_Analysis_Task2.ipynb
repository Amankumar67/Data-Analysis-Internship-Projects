```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# step for cleaned dataset

df = pd.read_csv("netflix_cleaned_dataset.csv")
df.head()
```

```python
# step for total number of Movies and TV Shows 

df["type"].unique()
```

```python
# step for total movies and shows 
content_count = df["type"].value_counts()
content_count
```

```python
# content percentage 

content_percentage = df["type"].value_counts(normalize=True) * 100
print("content percentage:" , content_percentage.round(2))
```

```python
 # Bar Chart - Movies vs TV Shows 

plt.figure(figsize=(4,5))

ax = sns.barplot(x = content_count.index,y= content_count.values)
plt.title("Netflix Content Type Distribution")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

```python
# pie chart = Content proportion 

plt.pie(content_count.values, labels= content_count.index, autopct="%1.1f%%", startangle=90)
plt.title("Netflix Movies vs TV Shows")
plt.show()
```

```python
 # step for Compare proportions 

# calculate the percentage 

content_summary = pd.DataFrame({"count": df["type"].value_counts(),
                                "percentage": df["type"].value_counts(normalize=True) * 100})

content_summary["percentage"] = content_summary["percentage"].round(2)

print(content_summary)
```

```python
 # create a comparison chart 

ax = sns.barplot(x = content_summary.index, y = content_summary["percentage"])

plt.title("Netflix Content proportion")
plt.xlabel("Content Type")
plt.ylabel("percentage (%)")
for container in ax.containers:
    ax.bar_label(container, fmt = "%.2f%%")

plt.show()
```

```python
 # calculate the difference 
movie_count = content_summary.loc["Movie","count"]
tv_show_count = content_summary.loc["TV Show", "count"]

difference = movie_count - tv_show_count
print("difference between Movies and TV Shows ", difference)
```

```python
 # calculate the movie-to-show ratio
ratio = movie_count / tv_show_count
print("Movies-to-TV Show ratio:", round(ratio, 2))
```

# step = summaries key finding 

Key Finding = The Analysis shows that movies make up a large proportion of the Netflix dataset than TV Shows. There is a difference opf 3,462 titles between the two content types , with approximately 2.3 Movies available for every 1 TV Show.
